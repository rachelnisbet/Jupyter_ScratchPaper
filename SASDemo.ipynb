# Read Data

```sas
libname demosas xlsx "~/Courses/PGVY01/data/orders.xlsx";
```

# Look at snapshot of data

```sas
proc print data=demosas.orders(obs=10);
run;
```

# Summary Statistics

```sas
proc means data=demosas.orders;
run;
```

# Frequencies

```sas
proc freq data=demosas.orders(obs=100);
    tables customer_group;
run;
```

# Convert data

```sas
data orders_clean;
set demosas.orders(obs=10);
cust_id = put(customer_ID, 8.);
run;
```

```sas
* Check;
proc contents data=orders_clean;
run;
```

# Creating data

```sas
data example_data;
    length First_Name $ 25 Last_Name $ 25;     
    input First_Name $ Last_Name $ age height;
    datalines;
    Rachel Nisbet 26 63
    Bruce Wayne 35 70
    Will Smith 52 74
    Oprah Winfrey 67 67
;
run;

proc print data=example_data;
run;
```
